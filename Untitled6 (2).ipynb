```python
from datasets import load_dataset

dataset = load_dataset("cnn_dailymail", "3.0.0")

def format_example(example):
    return {
        "text": f"### Instruction:\nSummarize the following article:\n{example['article']}\n\n### Response:\n{example['highlights']}"
    }

train_data = dataset["train"].map(format_example).select(range(500))   # small subset for Colab
val_data   = dataset["validation"].map(format_example).select(range(100))
```

```python
from peft import LoraConfig
from transformers import TrainingArguments
from trl import SFTTrainer

peft_config = LoraConfig(
    r=8,
    lora_alpha=32,
    lora_dropout=0.1,
    bias="none",
    task_type="CAUSAL_LM"
)

training_args = TrainingArguments(
    output_dir="/content/gemma2b-cnn-lora",
    per_device_train_batch_size=1,
    gradient_accumulation_steps=16,   # effective larger batch
    learning_rate=2e-4,
    num_train_epochs=1,
    fp16=True,
    save_strategy="epoch",
    eval_strategy="no",              # 🔥 disable eval to save VRAM
    logging_steps=50,
    report_to="none",
    gradient_checkpointing=True,
)

trainer = SFTTrainer(
    model=model,
    train_dataset=train_data,
    eval_dataset=None,        # 🔥 skip validation to prevent OOM
    peft_config=peft_config,
    args=training_args
)

trainer.train()
```

```python
trainer.push_to_hub("harintharan/gemma2b-cnn-lora-summarizer")
```
